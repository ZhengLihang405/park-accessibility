```python
import geopandas as gpd
import matplotlib.pyplot as plt
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

base = r"D:\code\data\shape"
landuse = gpd.read_file(base + r"\landuse.shp")
buildings = gpd.read_file(base + r"\buildings.shp")
natural = gpd.read_file(base + r"\natural.shp")

print(landuse.columns.tolist())
print(landuse["type"].value_counts().head(15))
print(len(buildings), "栋建筑")
```

```python
leisure = gpd.read_file(base + r"\leisure.shp")
print(leisure["type"].value_counts().head(10))
```

```python
# 公园绿地：休闲用地 + 草地 + 花坛
green_types = ["recreation_ground", "grass", "flowerbed"]
parks = landuse[landuse["type"].isin(green_types)].copy()
parks = parks.to_crs(parks.estimate_utm_crs())
print(len(parks), "个公园绿地")

# 500 m 服务范围
service = parks.buffer(500).union_all()
service = gpd.GeoSeries([service], crs=parks.crs)
print("服务范围面积(km²):", round(service.area.iloc[0] / 1e6, 2))

# 建筑覆盖率
pts = buildings.to_crs(parks.crs).copy()
pts["geometry"] = pts.centroid
pts["covered"] = pts.within(service.iloc[0])
print("建筑覆盖率:", round(pts["covered"].mean() * 100, 1), "%")
```

```python
fig, ax = plt.subplots(figsize=(10, 10))
service.plot(ax=ax, color="#a8d5a2", alpha=0.6)
parks.plot(ax=ax, color="#2e7d32")
ax.set_title("厦门本岛公园绿地 500m 服务范围", fontsize=14)
ax.axis("off")
plt.savefig(r"D:\code\park_service.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 10))
service.plot(ax=ax, color="#a8d5a2", alpha=0.4)
pts[pts["covered"]].plot(ax=ax, color="#9e9e9e", markersize=1, label="已覆盖")
pts[~pts["covered"]].plot(ax=ax, color="#e53935", markersize=1, label="盲区")
ax.set_title("建筑公园服务覆盖情况（红色为盲区）", fontsize=14)
ax.legend(markerscale=6)
ax.axis("off")
plt.savefig(r"D:\code\blind_spots.png", dpi=300, bbox_inches="tight")
plt.show()
```
